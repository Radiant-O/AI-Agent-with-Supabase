```python
# Dynamic and Static Typing

score = 10 # int
print(type(score))

score = 'name' # string
print(type(score))

# Numbers and Math Operators

# Addition (+), Subtraction (-), Multiplication (*): These are mainly the same way they operates.

# Division (/) : : These will always return float even if the result has no reminder
print (8 / 2)

# Floor Division (//) : These will always return a whole number even if the result has a reminder
print (11 // 2)
print (5 * 2.0)

# Exponetial (**)
print(2 ** 3)

# Modulus (%) : It will always return the reminder
print (8 % 5)


# Order of Operation In Python

# 1. Exponential (**)
# 2. Multiplication (*) and Division (/)
# 3. Addition (+) and Subtraction (-)

print(2 + 4 * 2 ** 3)

print(18 % 3, 15 / 3, 16 // 3, 2 ** 3)

print((3 * 2) + 2 **4)

print(1_000_232_454)
# is the same as
print(1000232454)

#The first one enhance readability
```

```python
# Python Key Operators: Comparision, Assignment, Identity Operators

# Equals (=)
a = 5

# Plus Equals (+=)
a += 2 # same as a = a + 2
print(a)

# Minus Equals (-=)
a -= 2 # same as a = a - 2
print(a)

# Star Equals (*=)
a *= 2 # same as a = a * 2
print(a)

# Slash Equals (/=)
a /= 2 # same as a = a / 2
print(a)

# Double stars Equals (**=)
a **= 2 # same as a = a ** 2
print(a)

# Percent Equals (%=)
a %= 2 # same as a = a % 2
print(a)

# Comparision Operators
# Equal to (==)
# Not Equal to (!=)
# Greater Than (>)
# Greater Than or equal to (>=)
# Less than (<)
# Less than or equal to (<=)

# Mutability and Immutability
# Mutable objects can be changed after they are created, while Immultabe objects can't

# Integers and Strings are Immutable items, they can't be change if it is change they change adress

a = 2
print(id(a))

a += 3
print(id(a))

# Lists, Tuple, Sets are mutable and they can be change

numbers = [1, 2, 3]
print(id(numbers))
numbers.append(10)
print(id(numbers))
```
